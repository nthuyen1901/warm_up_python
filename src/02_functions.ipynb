# Functions
A function is a named, reusable block of code designed to perform a specific task.

For example, we may need to perform a data cleaning step on every row in a large data set. Instead of spelling out this step over and over again, we write it once and then call it as we process each line.

## Defining and Calling Functions
In Python, functions are defined using the `def` keyword. Every function definition consists of two parts:

- **The signature**: The name of the function and its input arguments
- **The body**: The code block that is executed when the function is called

````python
# Defining a function
def greet():
    print("hello world")
````

````python
# Calling a function
greet()
````

### Code Blocks
A code block is a unit of code that is executed together. In many programming languages, code blocks are enclosed by curly braces `{}`. In Python, a code block is defined by **indentation**, usually four spaces. One level of indentation indicates one coherend block of logic.


#### Python

### Arguments and Return Values
To make a function useful, we often send it data (arguments) and get a result back (return value). Some important terminology to keep in mind:

- **Parameters:** Placeholders for data defined in the signature of the function.
- **Arguments:** The actual data passed to the function when calling it.
- **Return:** Use the `return` keyword to send data back to the main program.

Note that the `return` keyword always terminates the function's execution.

````python
def add_one(num):
    incremented_num = num + 1
    return incremented_num
````

````python
my_num = input("Enter a number: ")
my_num = int(my_num)
print(add_one(my_num))
````

Python offers flexibility in how you can pass arguments to functions:

- **Positional arguments** are the default. They are matched by the order they appear.
- **Keyword arguments** are explicitly named parameters.
- **Optional arguments** have a default value that is used when they are absent in a function call.

````python
def multiply(x, y):
    return x * y
````

````python
def area(height, width):
    return height * width
````

````python
def exponentiate(base, exponent=2):
    return base ** exponent
````

````python
# Calling the multiply() functions with positional arguments
multiply(4,2)
````

````python
# Calling the multiply() function with keyword arguments
multiply(y=2, x=4)
````

````python
# Calling area() with keyword arguments
area(width=2, height=4)
````

````python
# Calling area() with positional arguments
area(2, 4)
````

````python
# Calling exponentiate() with one argument
exponentiate(3)
````

````python
# Calling exponentiate() with two arguments
exponentiate(3, 3)
````

## Scope: Local vs Global
Think of functions as programs inside a program. When a function is called, the execution of the main program is halted until the function finishes execution.

One important consequence is that variables defined inside a function only exist within that function. They are destroyed when the function finishes execution. The `return` keyword is necessary to make this data available to the main function.

```python
def increment(num, increment_by = 1):
    incremented = num + increment_by  # Local scope: Only available in the function
    return incremented

my_incremented_number = increment(4, 2)
print(my_incremented_number)  # 6
print(incremented)            # Error!
```

Variables assigned outside your function - as part of your main program - are available everywhere. You can read data in a function that is not passed as an argument. You can even manipulate it, if you use the `global` keyword inside your function. However, this is a frequent source of bugs and generally not recommended, unless you *really* know what you're doing!

## Readability
Code is always easier to write than to read. To do your fellow programmers and your future self a favour, aim to make your code as clean and readable as possible. To that end you can:

1. Choose clear names for variables and functions, and remain consistent in your naming strategy
2. Document your function using [Docstrings](https://peps.python.org/pep-0257/)
3. Use [type hints](https://docs.python.org/3/library/typing.html) for runtime support

### Naming
Picking the "right" name for a variable or a function becomes surprisingly hard once you're tackling a complex problem in code. A good rule of thumb is to express intent through the name. And more importantly, stay consistent with your approach: If you create a function that assembles text in a specific way with the `make_` prefix, do the same for all future functions that do similar work.

```python
# Bad example: Undescriptive variable names and inconsistent function naming
a = "hello"
b = "goodbye"
c = input()

def say_bye(b):
    return f"{b} {c}"

def make_greeting(a, b):
    return f"{a} {b}"
```

```python
# Better
welcome_message = "hello"
goodbye_message = "bye"
name = input()

def make_greeting(name):
    return f"{welcome_message} {name}"

def make_goodbye(name):
    return f"{goodbye_message} {name}"
```

### Docstrings
A Python standard to document functions. Docstrings are wrapped in triple quotation marks `""" ... """`and written between the signature and body of a function. They can be on a single line or cover multiple lines. Code editors such as Visual Studio code read docstrings and use them to display hints in the editor itself.

```python
def calculate_total(base_price, tax_rate=0.20):
    """Calculate the total price including VAT.
    
    Arguments:
        base_price: The item price excluding VAT.
        tax_rate: The tax percentage to add. Defaults to 0.2 (20%).

    Returns:
        Final price: Sum of base price and tax.
    """
    return base_price + (base_price * tax_rate)
```

### Type Hints
An option to enhance the documentation of your function. As the name suggests, type hints are merely _hints_ that tell what data type a function expects to receive as input, and what return type the rest of the program can expect from the function. Do not confuse them with static type declarations from other programming languages.

Like docstrings, type hints are read by code editors and used to display enhanced tooltips as well as warnings when you try to pass an unexpected type to a function.

```python
def calculate_total(base_price: float, tax_rate: float = 0.2) -> float:
    """Calculate the total price including VAT. Default tax rate is 20%."""
    return base_price + (base_price * tax_rate)
```

What does this function do? Can you rewrite it to make it easier to understand?

```python
def d(x1, x2, y1, y2):
    return abs(x1 - x2) + abs(y1 - y2)
```

````python
# Write a more readable version of this function
def manhattan_distance(x1, y1, x2, y2):
    horizontal_distance = abs(x1 - x2)
    vertical_distance = abs(y1 - y2)

    return horizontal_distance + vertical_distance
````
